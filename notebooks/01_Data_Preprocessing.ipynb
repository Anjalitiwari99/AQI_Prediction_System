## Extracting Tables from PDF

```python
import pdfplumber

pdf_path = "data/prayagraj_air_quality_data (1).pdf"

all_tables = []

with pdfplumber.open(pdf_path) as pdf:
    print(f"Total Pages: {len(pdf.pages)}")

    for page_num, page in enumerate(pdf.pages):

        table = page.extract_table()

        if table:
            all_tables.append(table)
            print(f"Page {page_num+1}: Table Found ✅")

        else:
            print(f"Page {page_num+1}: No Table ❌")
```

```python
import pdfplumber

pdf_path = "../data/prayagraj_air_quality_data (1).pdf"

all_tables = []

with pdfplumber.open(pdf_path) as pdf:
    print(f"Total Pages: {len(pdf.pages)}")

    for page_num, page in enumerate(pdf.pages):
        table = page.extract_table()

        if table:
            all_tables.append(table)
            print(f"Page {page_num+1}: Table Found ✅")
        else:
            print(f"Page {page_num+1}: No Table ❌")
```

## Preview Extracted Tables

```python
print("Total Tables Extracted:", len(all_tables))

print("\nFirst Table Preview:\n")

for row in all_tables[0][:10]:
    print(row)
```

## Converting Extracted Table into DataFrame

```python
# Convert first extracted table into DataFrame

df = pd.DataFrame(all_tables[0])

print("Shape of DataFrame:", df.shape)

df.head(10)
```

```python
print("Shape :", df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

## Setting Proper Column Names

```python
df.columns = [
    "Parameter",
    "Civil Lines",
    "Rambagh",
    "Naini",
    "Phoolpur",
    "Katra",
    "Teliyarganj",
    "Company Bagh"
]

print("Column names updated successfully ✅")
```

## Removing Unnecessary Header Rows

```python
# Keep only actual air quality measurements
df_clean = df.iloc[4:].reset_index(drop=True)

print("Cleaned Shape:", df_clean.shape)

df_clean.head()
```

## Transposing Dataset

```python
df_final = df_clean.set_index("Parameter").T

df_final
```

## Resetting Index and Creating Location Column

```python
df_final = df_final.reset_index()

df_final = df_final.rename(columns={"index": "Location"})

df_final.head()
```

```python
df_final.columns
```

## Checking Final Dataset Information

```python
df_final.info()
```

## Converting Data Types

```python
# Convert all columns except Location to numeric

numeric_columns = df_final.columns[1:]

df_final[numeric_columns] = df_final[numeric_columns].apply(pd.to_numeric)

print("Data types converted successfully ✅")
```

```python
df_final.info()
```
